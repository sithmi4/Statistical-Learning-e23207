## Q1. Statistical Mechanics of the Quantum Oscillator
### 1. The Partition Function Derivation
Given the energy eigenvalues:
$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad j = 0, 1, 2, \dots$$

The canonical partition function is defined as:
$$Z(T) = \sum_{j=0}^\infty e^{-\frac{E_j}{k_B T}} = \sum_{j=0}^\infty e^{-\frac{\hbar \omega (j + 1/2)}{k_B T}}$$

We can factor out the constant term corresponding to $j=0$ (the zero-point energy):
$$Z(T) = e^{-\frac{\hbar \omega}{2k_B T}} \sum_{j=0}^\infty \left(e^{-\frac{\hbar \omega}{k_B T}}\right)^j$$

Let $r = e^{-\frac{\hbar \omega}{k_B T}}$. Since $\hbar, \omega, k_B, T > 0$, we have $0 < r < 1$. Thus, the summation is a convergent infinite geometric series with a sum equal to $\frac{1}{1-r}$:
$$\sum_{j=0}^\infty r^j = \frac{1}{1 - e^{-\frac{\hbar \omega}{k_B T}}}$$

Substituting this back yields the desired compact closed form:
$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


### 2. Thermodynamic Potentials
* **Helmholtz Free Energy ($F$):**
$$F(T) = -k_B T \ln Z = -k_B T \ln \left( \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}} \right)$$
Using logarithm properties:
$$F(T) = -k_B T \left[ -\frac{\hbar \omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right] = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)$$

* **Internal Energy ($U$):**
Let $\beta = \frac{1}{k_B T}$. We rewrite $Z$ in terms of $\beta$:
$$Z(\beta) = \frac{e^{-\frac{\beta \hbar \omega}{2}}}{1 - e^{-\beta \hbar \omega}}$$
$$\ln Z = -\frac{\beta \hbar \omega}{2} - \ln\left(1 - e^{-\beta \hbar \omega}\right)$$
Differentiating with respect to $\beta$:
$$U = -\frac{\partial \ln Z}{\partial \beta} = -\left[ -\frac{\hbar \omega}{2} - \frac{1}{1 - e^{-\beta \hbar \omega}} \cdot \left(-e^{-\beta \hbar \omega}\right) \cdot (-\hbar \omega) \right]$$
$$U = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}} = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1}$$
Substituting back $\beta = \frac{1}{k_B T}$:
$$U(T) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$


### 3. Entropy Calculation and Verification
Using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$:
$$F(T) = \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right)$$
Using the product rule for the derivative with respect to $T$:
$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) - k_B T \cdot \frac{1}{1 - e^{-\frac{\hbar \omega}{k_B T}}} \cdot \left(-e^{-\frac{\hbar \omega}{k_B T}}\right) \cdot \left(\frac{\hbar \omega}{k_B T^2}\right)$$
$$S = -k_B \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{T} \frac{1}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$

**Verification of $U = F + TS$:**
$$TS = -k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1}$$
Adding $F$ and $TS$:
$$F + TS = \left[ \frac{\hbar \omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) \right] + \left[ -k_B T \ln\left(1 - e^{-\frac{\hbar \omega}{k_B T}}\right) + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} \right]$$
$$F + TS = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\frac{\hbar \omega}{k_B T}} - 1} = U(T)$$
The identity holds perfectly.


### 4. Heat Capacity and Hyperbolic Form
The heat capacity is $C_V = \frac{\partial U}{\partial T}$:
$$U(T) = \frac{\hbar \omega}{2} + \hbar \omega \left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^{-1}$$
Using the chain rule:
$$C_V = \hbar \omega \cdot (-1)\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^{-2} \cdot e^{\frac{\hbar \omega}{k_B T}} \cdot \left(-\frac{\hbar \omega}{k_B T^2}\right) = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar \omega}{k_B T}}}{\left(e^{\frac{\hbar \omega}{k_B T}} - 1\right)^2}$$

To rewrite in hyperbolic terms, multiply the numerator and denominator by $e^{-\frac{\hbar \omega}{k_B T}}$:
$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{\left(e^{\frac{\hbar \omega}{2k_B T}} - e^{-\frac{\hbar \omega}{2k_B T}}\right)^2}$$
Since $\sinh(x) = \frac{e^x - e^{-x}}{2} \implies 2\sinh(x) = e^x - e^{-x}$:
$$C_V = k_B \left(\frac{\hbar \omega}{k_B T}\right)^2 \frac{1}{\left(2\sinh\left(\frac{\hbar \omega}{2k_B T}\right)\right)^2} = k_B \left(\frac{\hbar \omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$


### 5. Asymptotic Behavior (Discussion)
In the high-temperature limit where $k_B T \gg \hbar \omega$, let $x = \frac{\hbar \omega}{2k_B T} \to 0$.
Using the Taylor expansion for small arguments, $\sinh(x) \approx x$.
Substituting this approximation into our $C_V$ formula:
$$C_V \approx k_B x^2 \cdot \frac{1}{x^2} = k_B$$

**Comparison to Equipartition Theorem:**
A classical 1D harmonic oscillator has two degrees of freedom (one kinetic energy, one potential energy). According to the classical Equipartition Theorem, each quadratic degree of freedom contributes $\frac{1}{2}k_B$ to the heat capacity, yielding a total classical value of $C_V = k_B$.
Therefore, our quantum-mechanical derivation matches the classical equipartition theorem prediction exactly in the high-temperature limit.


#### Symbolic Python Verification for the Quantum Oscillator:


```python
import sympy as sp

# Define symbols
hbar, omega, kB, T, beta = sp.symbols('hbar omega k_B T beta', positive=True)

# 1. Define Partition Function in terms of beta
Z = sp.exp(-beta * hbar * omega / 2) / (1 - sp.exp(-beta * hbar * omega))

# 2. Derive Internal Energy
U = -sp.diff(sp.log(Z), beta)
U = U.subs(beta, 1 / (kB * T))
print("Internal Energy U(T):")
sp.pprint(sp.simplify(U))

# 3. Derive Heat Capacity CV
CV = sp.diff(U, T)
print("\nHeat Capacity CV(T):")
sp.pprint(sp.simplify(CV))
```

## Q2. The Maximum Entropy Distribution under Mean and Variance Constraints
We want to maximize the continuous Shannon entropy functional:
$$H(p) = -\int_{-\infty}^{\infty} p(x) \ln p(x) dx$$
Subject to the constraints:
1. Normalized Probability: $\int_{-\infty}^{\infty} p(x) dx = 1$
2. Mean Constraint: $\int_{-\infty}^{\infty} x p(x) dx = \mu$
3. Variance Constraint: $\int_{-\infty}^{\infty} (x-\mu)^2 p(x) dx = \sigma^2$

Using the method of Lagrange multipliers, we formulate the functional objective function $\mathcal{L}$:
$$\mathcal{L}(p, \lambda_0, \lambda_1, \lambda_2) = -\int_{-\infty}^{\infty} p(x) \ln p(x) dx - \lambda_0 \left(\int_{-\infty}^{\infty} p(x) dx - 1\right) - \lambda_1 \left(\int_{-\infty}^{\infty} x p(x) dx - \mu\right) - \lambda_2 \left(\int_{-\infty}^{\infty} (x-\mu)^2 p(x) dx - \sigma^2\right)$$

Taking the functional derivative with respect to $p(x)$ and setting it to zero:
$$\frac{\delta \mathcal{L}}{\delta p(x)} = -\ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$
Solving for $p(x)$:
$$p(x) = e^{-1 - \lambda_0} e^{-\lambda_1 x - \lambda_2 (x - \mu)^2}$$

To satisfy the symmetric expectation values centered around $\mu$, the linear coefficient $\lambda_1$ must equal $0$. Let $C = e^{-1 - \lambda_0}$ represent our normalizing constant:
$$p(x) = C e^{-\lambda_2 (x - \mu)^2}$$

Using the standard Gaussian normalization integral $\int_{-\infty}^{\infty} e^{-a(x-\mu)^2}dx = \sqrt{\frac{\pi}{a}}$:
$$C \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies C = \sqrt{\frac{\lambda_2}{\pi}}$$

Now we enforce the variance constraint:
$$\int_{-\infty}^{\infty} (x-\mu)^2 \sqrt{\frac{\lambda_2}{\pi}} e^{-\lambda_2 (x - \mu)^2} dx = \sigma^2$$
Using the standard identity $\int_{-\infty}^{\infty} y^2 e^{-ay^2}dy = \frac{1}{2a}\sqrt{\frac{\pi}{a}}$:
$$\sqrt{\frac{\lambda_2}{\pi}} \cdot \left( \frac{1}{2\lambda_2} \sqrt{\frac{\pi}{\lambda_2}} \right) = \sigma^2 \implies \frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$

Substituting $\lambda_2$ back into our equations yields our exact solution, which is the Gaussian distribution:
$$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$


## Q3. Multiclass Logistic Regression: Automatic Classification of Emails
We have 5 known text feature states with the following historical probabilities:
* $P(\mathbf{x}_A) = 0.40, \quad \mathbf{x}_A = (1, 0, 0)^T$
* $P(\mathbf{x}_B) = 0.25, \quad \mathbf{x}_B = (0, 1, 0)^T$
* $P(\mathbf{x}_C) = 0.05, \quad \mathbf{x}_C = (0, 0, 1)^T$
* $P(\mathbf{x}_D) = 0.15, \quad \mathbf{x}_D = (0, 1, 1)^T$
* $P(\mathbf{x}_E) = 0.05, \quad \mathbf{x}_E = (1, 1, 0)^T$

The empirical constraints on the target destination $Y=1$ (Technical support) are given as:
1. $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = 0.05$
2. $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = 0.15$
3. $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = 0.16$

In a Maximum Entropy classifier framework (equivalent to multinomial logistic regression), the conditional probability takes a Softmax distribution form. Here we compute the required conditional parameters numerically to satisfy these constraints.


```python
import numpy as np
from scipy.optimize import minimize

# Setup the data structures for the 5 active observed scenarios
X_matrix = np.array([
    [1, 0, 0], # A
    [0, 1, 0], # B
    [0, 0, 1], # C
    [0, 1, 1], # D
    [1, 1, 0]  # E
])
P_x = np.array([0.40, 0.25, 0.05, 0.15, 0.05])
target_expectations = np.array([0.05, 0.15, 0.16])

def loss_function(weights):
    # compute conditional probability P(Y=1 | x) = sigmoid(w . x)
    logits = np.dot(X_matrix, weights)
    P_y1_given_x = 1 / (1 + np.exp(-logits))

    # Compute expectations: sum_x P(x) * x_i * P(Y=1 | x)
    computed_expectations = np.zeros(3)
    for i in range(3):
        computed_expectations[i] = np.sum(P_x * X_matrix[:, i] * P_y1_given_x)

    # Return Mean Squared Error from targets
    return np.sum((computed_expectations - target_expectations)**2)

# Solve for the optimal weights using optimization
initial_guess = np.zeros(3)
res = minimize(loss_function, initial_guess, method='BFGS')
optimal_weights = res.x

print("Converged Optimal Weights (w1, w2, w3):")
print(optimal_weights)

# Verification printout
logits = np.dot(X_matrix, optimal_weights)
final_P = 1 / (1 + np.exp(-logits))
print("\nVerified expectations vs Targets:")
for i in range(3):
    val = np.sum(P_x * X_matrix[:, i] * final_P)
    print(f"Feature x{i+1}: Computed={val:.3f} | Target={target_expectations[i]:.3f}")
```

## Q4. Asset Pricing and Portfolio Optimization
We are seeking the MaxEnt probability density function $p(S_T)$ over $S_T \in (0, \infty)$ under the following conditions:
1. $\mathbb{E}[1] = \int_0^\infty p(S_T) dS_T = 1$
2. $\mathbb{E}[S_T] = \int_0^\infty S_T p(S_T) dS_T = 102$
3. $\mathbb{E}[S_T^2] = \int_0^\infty S_T^2 p(S_T) dS_T = 10,600$
4. $\mathbb{E}[\ln(S_T)] = \int_0^\infty \ln(S_T) p(S_T) dS_T = 4.60$

The general Gibbs form for the maximum entropy density is:
$$p(S_T) = \exp\left(-\lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)\right) = \frac{1}{Z} S_T^{-\lambda_3} e^{-\lambda_1 S_T - \lambda_2 S_T^2}$$

We use numerical quadrature routines (`scipy.integrate.quad`) alongside a multi-dimensional root finder (`scipy.optimize.root`) to isolate the unique Lagrange multiplier parameters.


```python
import numpy as np
from scipy.integrate import quad
from scipy.optimize import root
import matplotlib.pyplot as plt

# Target boundary constraints
targets = np.array([1.0, 102.0, 10600.0, 4.60])

def compute_moments(lambdas):
    # lambdas corresponds to [lambda_1, lambda_2, lambda_3]
    l1, l2, l3 = lambdas

    # Partition function integrand
    def integrand_Z(s):
        # Guard against numerical underflow / overflow overflows safely
        val = -l1*s - l2*(s**2) - l3*np.log(s)
        return np.exp(val)

    # Numerical integration for normalization factor Z
    Z, _ = quad(integrand_Z, 1e-3, 500, limit=100)

    # Define expectation integrands
    def integrand_s1(s): return s * integrand_Z(s) / Z
    def integrand_s2(s): return (s**2) * integrand_Z(s) / Z
    def integrand_log(s): return np.log(s) * integrand_Z(s) / Z

    E_1 = 1.0
    E_s, _ = quad(integrand_s1, 1e-3, 500, limit=100)
    E_s2, _ = quad(integrand_s2, 1e-3, 500, limit=100)
    E_log, _ = quad(integrand_log, 1e-3, 500, limit=100)

    return np.array([E_1, E_s, E_s2, E_log]), Z

def system_to_solve(lambdas):
    moments, _ = compute_moments(lambdas)
    # We want calculated moments to match target values 1, 2, and 3
    return moments[1:] - targets[1:]

# Initial robust guess for multipliers [lambda_1, lambda_2, lambda_3]
initial_lambdas = np.array([0.01, 0.0001, -0.5])
sol = root(system_to_solve, initial_lambdas, method='lm')
opt_lambdas = sol.x

# Compute final values and partition function Z
final_moments, final_Z = compute_moments(opt_lambdas)

print("Converged Lagrange Multipliers:")
print(f"  lambda_0 (ln Z): {np.log(final_Z):.4f}")
print(f"  lambda_1 (Price): {opt_lambdas[0]:.6f}")
print(f"  lambda_2 (Variance): {opt_lambdas[1]:.6f}")
print(f"  lambda_3 (Skew Log): {opt_lambdas[2]:.6f}")

# Generate the non-Gaussian asset risk distribution graph
s_vals = np.linspace(50, 160, 500)
p_vals = [np.exp(-opt_lambdas[0]*s - opt_lambdas[1]*(s**2) - opt_lambdas[2]*np.log(s)) / final_Z for s in s_vals]

plt.figure(figsize=(9, 5))
plt.plot(s_vals, p_vals, 'g-', lw=2.5, label='MaxEnt Fitted Asset Distribution')
plt.axvline(102, color='r', linestyle='--', label='Expected Forward Price = $102')
plt.title('Non-Gaussian MaxEnt Asset Price Risk Profile')
plt.xlabel('Future Price Asset S_T ($)')
plt.ylabel('Probability Density')
plt.grid(True, alpha=0.4)
plt.legend()
plt.show()
```
